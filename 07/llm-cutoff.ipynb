```python
from transformers import pipeline
qa = pipeline(task="text-generation", model="Qwen/Qwen3-0.6B", do_sample=False, temperature=0, truncation=True)

prompts  = ["Who is Ramanujan? /nothink", "Who won the 2024 US presidential election? /nothink"]
for prompt in prompts:
    print(qa(prompt, max_new_tokens=100)[0]['generated_text'] + "\n-----------\n")
```

```python
from openai import OpenAI
client = OpenAI()
# sometimes a fast way to find-out the knowledge cutoff of a model
# diffrence between correctness and helpfulness of the answers
prompt = "Who won the 2024 US presidential election?"
for model in ["gpt-5-nano","gpt-4.1-mini", "gpt-3.5-turbo"]:
    print(f"Model: {model}")
    response = client.responses.create(
        model=model,
        input=prompt
    )
    print(response.output_text)
    print("\n-----------\n")
```
