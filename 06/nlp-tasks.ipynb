````python
# masked language modeling
from transformers import pipeline

unmasker = pipeline("fill-mask", model="bert-base-uncased")
_ = unmasker("This cooking course will teach you a lot of [MASK] recipes.", top_k=2)

display(_)
print("\n---\n")
unmasker.model  # model architecture (enconder with MLM head)
````

````python
# named entity recognition
from transformers import pipeline

ner = pipeline("ner", model="dbmdz/bert-large-cased-finetuned-conll03-english", grouped_entities=True)
_ = ner("My name is Massimo and I work at Websolute in Italy.")
display(_)

print("\n---\n")
ner.model  # model architecture (encoder with self-attention)
````

````python
# feature extraction: extract vector representations of input text == embeddings
from transformers import pipeline
import torch
feature_extractor = pipeline("feature-extraction", model="distilbert-base-uncased-finetuned-sst-2-english")
_ = feature_extractor("The book was great, the movie not so much!")
_ = torch.tensor(_)
display(_.shape)  # sequence length, hidden size

print("\n---\n")
print(feature_extractor.model)  # model architecture
````

````python
# sentiment analysis
from transformers import pipeline

classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
_one = classifier("I've been waiting for an cooking course my whole life.")
print(_one)

print("\n")
_many = classifier(["I love programming.", "I hate bugs."])
display(_many)

print("\n---\n")
print(classifier.model)  # model architecture
````

````python
# zero-shot classification
from transformers import pipeline

classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
_one = classifier(
    "This is a course about the Transformers library",
    candidate_labels=["education", "politics", "business"],
)
print(_one)

print("\n")

_many = classifier(
    [
        "I want to learn about AI",
        "The government has announced new policies",
        "The stock market is experiencing volatility",
    ],
    candidate_labels=["education", "politics", "business"],
)
display(_many)

print("\n---\n")
classifier.model  # model architecture
````

````python
# text generation
from transformers import pipeline
generator = pipeline("text-generation", model="Qwen/Qwen3-0.6B", trust_remote_code=True)
_one = generator("In this AI course, we are learning how to", max_new_tokens=100, num_return_sequences=2)

display(_one)

print("\n---\n")
generator.model  # model architecture (decoder with lm head)
````

````python
# question answering
from transformers import pipeline

question_answerer = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")
_ = question_answerer(
    question="Where do I work?",
    context="My name is Massimo and I work at Websolute in Italy.",
)

display(_)

print("\n---\n")
question_answerer.model  # model architecture (encoder with attention)
````

````python
# part-of-speech tagging
from transformers import pipeline
post = pipeline("token-classification", model="vblagoje/bert-english-uncased-finetuned-pos", grouped_entities=True)
_ = post("My name is Massimo and I work at Websolute in Italy.")
display(_)

print("\n---\n")
post.model  # model architecture (encoder with self-attention)
````

````python
# translation
from transformers import pipeline

_test = "Ciao, mi chiamo Massimo e lavoro presso Websolute in Italia."
translator = pipeline("translation", model="Helsinki-NLP/opus-mt-ROMANCE-en")
_ =translator(_test, max_length=40)
display(_)
print("\n---\n")
translator.model # model architecture (encoder-decoder with lm head)
````

````python
# summarization
from transformers import pipeline

summarizer = pipeline("summarization", model="facebook/bart-large-cnn")
_ = summarizer(
    """
    America has changed dramatically during recent years. Not only has the number of 
    graduates in traditional engineering disciplines such as mechanical, civil, 
    electrical, chemical, and aeronautical engineering declined, but in most of 
    the premier American universities engineering curricula now concentrate on 
    and encourage largely the study of engineering science. As a result, there 
    are declining offerings in engineering subjects dealing with infrastructure, 
    the environment, and related issues, and greater concentration on high 
    technology subjects, largely supporting increasingly complex scientific 
    developments. While the latter is important, it should not be at the expense 
    of more traditional engineering.

    Rapidly developing economies such as China and India, as well as other 
    industrial countries in Europe and Asia, continue to encourage and advance 
    the teaching of engineering. Both China and India, respectively, graduate 
    six and eight times as many traditional engineers as does the United States. 
    Other industrial countries at minimum maintain their output, while America 
    suffers an increasingly serious decline in the number of engineering graduates 
    and a lack of well-educated engineers.
"""
)
display(_)
print("\n---\n")
display(summarizer.model) # model architecture (encoder-decoder with attention)
````

````python
# Compare Encoder-Only, Decoder-Only, and Encoder-Decoder Models
from transformers import pipeline
import torch

print("=" * 100)
print("TRANSFORMER MODEL ARCHITECTURES COMPARISON")
print("=" * 100)

# 1. ENCODER-ONLY MODEL (BERT)
print("\n1️⃣  ENCODER-ONLY MODEL (BERT) - Fill Mask")
print("-" * 100)
encoder_only = pipeline("fill-mask", model="google-bert/bert-base-cased")
result_encoder = encoder_only("The capital of France is [MASK].", top_k=3)
print("Task: Fill in the [MASK] token")
print(f"Input: 'The capital of France is [MASK].'")
print(f"Output:")
for r in result_encoder:
    print(f"  - {r['sequence']} (score: {r['score']:.4f})")

print("\nModel Architecture:")
print(encoder_only.model)
print(f"\nTotal parameters: {sum(p.numel() for p in encoder_only.model.parameters()):,}")

# 2. DECODER-ONLY MODEL (Qwen)
print("\n\n2️⃣  DECODER-ONLY MODEL (Qwen) - Text Generation")
print("-" * 100)
decoder_only = pipeline("text-generation", model="Qwen/Qwen3-0.6B", trust_remote_code=True)
result_decoder = decoder_only(
    "The capital of France is",
    max_new_tokens=20,
    num_return_sequences=2,
    temperature=0.7
)
print("Task: Generate continuation of text")
print(f"Input: 'The capital of France is'")
print(f"Output:")
for i, r in enumerate(result_decoder, 1):
    print(f"  {i}. {r['generated_text']}")

print("\nModel Architecture:")
print(decoder_only.model)
print(f"\nTotal parameters: {sum(p.numel() for p in decoder_only.model.parameters()):,}")

# 3. ENCODER-DECODER MODEL (BART)
print("\n\n3️⃣  ENCODER-DECODER MODEL (BART) - Summarization")
print("-" * 100)
encoder_decoder = pipeline("summarization", model="facebook/bart-large-cnn")
text_to_summarize = """
The Eiffel Tower is a wrought-iron lattice tower on the Champ de Mars in Paris, France. 
It is named after the engineer Gustave Eiffel, whose company designed and built the tower. 
Constructed from 1887 to 1889 as the entrance to the 1889 World's Fair, it was initially 
criticized by some of France's leading artists and intellectuals for its design, but it has 
become a global cultural icon of France and one of the most recognizable structures in the world.
"""
result_enc_dec = encoder_decoder(text_to_summarize, max_length=50, min_length=20)
print("Task: Summarize long text")
print(f"Input: {text_to_summarize[:100]}...")
print(f"Output: {result_enc_dec[0]['summary_text']}")

print("\nModel Architecture:")
print(encoder_decoder.model)
print(f"\nTotal parameters: {sum(p.numel() for p in encoder_decoder.model.parameters()):,}")

# COMPARISON TABLE
print("\n\n" + "=" * 100)
print("COMPARISON TABLE")
print("=" * 100)

from IPython.display import Markdown, display

comparison_table = """
| Feature | Encoder-Only (BERT) | Decoder-Only (GPT like) | Encoder-Decoder (BART) |
|---------|---------------------|----------------------|------------------------|
| **Architecture** | Bidirectional self-attention | Unidirectional self-attention | Encoder + Decoder with cross-attention |
| **Input Processing** | Processes full sequence at once | Processes left-to-right only | Encoder processes input, decoder generates |
| **Attention Mechanism** | Can see all tokens (past & future) | Can only see previous tokens | Encoder: bidirectional, Decoder: causal |
| **Best For** | Understanding tasks | Generation tasks | Sequence-to-sequence tasks |
| **Common Tasks** | Classification, NER, Q&A, Fill-mask | Text generation, completion | Translation, summarization, paraphrasing |
| **Example Models** | BERT, RoBERTa, ALBERT | GPT-like, Llama | BART, T5, mBART |
| **Strengths** | Deep understanding of context | Creative text generation | Combines understanding + generation |
| **Weaknesses** | Cannot generate new text | No bidirectional context | More parameters, slower |
| **Training Objective** | Masked Language Modeling (MLM) | Causal Language Modeling (CLM) | Denoising autoencoding |
| **Key Component** | Self-attention layers | Causal self-attention + MLP | Cross-attention between encoder-decoder |
"""

display(Markdown(comparison_table))

# VISUAL COMPARISON
print("\n\n" + "=" * 100)
print("ARCHITECTURE DIAGRAMS")
print("=" * 100)

arch_diagram = """
### 🏗️ Architecture Comparison

**ENCODER-ONLY (BERT)**
```
Input Tokens
     ↓
[Embedding Layer]
     ↓
┌─────────────────┐
│ Transformer     │
│ Encoder Blocks  │  ← Bidirectional attention
│ (12-24 layers)  │     Can see ALL tokens
└─────────────────┘
     ↓
Output Representations
     ↓
[Task-Specific Head]
```

**DECODER-ONLY (GPT)**
```
Input Tokens
     ↓
[Embedding Layer]
     ↓
┌─────────────────┐
│ Transformer     │
│ Decoder Blocks  │  ← Causal/Masked attention
│ (12-48 layers)  │     Can only see PREVIOUS tokens
└─────────────────┘
     ↓
[Language Model Head]
     ↓
Generated Tokens
```

**ENCODER-DECODER (BART)**
```
Input Tokens
     ↓
[Embedding Layer]
     ↓
┌─────────────────┐      ┌─────────────────┐
│ Transformer     │      │ Transformer     │
│ Encoder Blocks  │──────│ Decoder Blocks  │  ← Cross-attention
│ (6-12 layers)   │      │ (6-12 layers)   │     from decoder to encoder
└─────────────────┘      └─────────────────┘
 ↑ Bidirectional          ↑ Causal
     ↓
[Output Head]
     ↓
Generated Sequence
```
"""

display(Markdown(arch_diagram))

# USE CASES
print("\n\n" + "=" * 100)
print("PRACTICAL USE CASES")
print("=" * 100)

use_cases = """
### 📋 When to Use Each Architecture

| Use Case | Recommended Architecture | Why? |
|----------|-------------------------|------|
| Sentiment Analysis | **Encoder-Only** | Need to understand full context |
| Named Entity Recognition | **Encoder-Only** | Need bidirectional context |
| Question Answering | **Encoder-Only** | Extract answers from context |
| Text Classification | **Encoder-Only** | Understand and categorize |
| Creative Writing | **Decoder-Only** | Generate coherent new text |
| Code Completion | **Decoder-Only** | Predict next tokens |
| Chatbots | **Decoder-Only** | Generate conversational responses |
| Translation | **Encoder-Decoder** | Transform one sequence to another |
| Summarization | **Encoder-Decoder** | Compress while maintaining meaning |
| Paraphrasing | **Encoder-Decoder** | Rewrite while preserving semantics |
"""

display(Markdown(use_cases))

# Clean up
del encoder_only, decoder_only, encoder_decoder
torch.cuda.empty_cache()
import gc
gc.collect()
````
