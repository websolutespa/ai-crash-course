````python
from mcp_server_time.server import TimeServer, get_local_tz
from datetime import datetime
server = TimeServer()
# public available methods
methods = []
for method in dir(server):
    if not method.startswith('_'):
        methods.append(method)
print(f"Methods: {methods}")

timezone = str(get_local_tz())
print(f"Local timezone {timezone}")
current_time = server.get_current_time(timezone)
print(f"{timezone} time: {current_time}")
print(f"UTC time: {server.get_current_time('UTC')}")
hh_mm = datetime.fromisoformat(current_time.datetime).strftime('%H:%M')
target_tz = 'America/New_York'
print(f"Time different with {target_tz} time: {server.convert_time(timezone, hh_mm, target_tz).time_difference}")

#see __main__.py & __init__.py for more details
````

````python
import subprocess
import json

# Start MCP server and send tools/list request
process = subprocess.Popen(
    ['uvx', 'mcp-server-time'],
    stdin=subprocess.PIPE, 
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE,
    text=True
)

# initialize request
init_request = {
    "jsonrpc": "2.0",
    "id": 1,
    "method": "initialize",
    "params": {
        "protocolVersion": "2025-11-25",
        "capabilities": {},
        "clientInfo": {"name": "test", "version": "1.0"}
    }
}

# List tools
tools_request = {
    "jsonrpc": "2.0",
    "id": 2,
    "method": "tools/list"
}

process.stdin.write(json.dumps(init_request) + '\n')
process.stdin.write(json.dumps(tools_request) + '\n')
process.stdin.flush()
process.stdin.close()

# read responses from stdout
for line in process.stdout:
    if line.strip():
        print(json.dumps(json.loads(line), indent=2))
````

````python
import subprocess
import json

# Python: Invoke get_current_time tool
process = subprocess.Popen(
    ['uvx', 'mcp-server-time'],
    stdin=subprocess.PIPE,
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE,
    text=True
)

# Initialize
init_request = {
    "jsonrpc": "2.0",
    "id": 1,
    "method": "initialize",
    "params": {
        "protocolVersion": "2025-11-25",
        "capabilities": {},
        "clientInfo": {"name": "test", "version": "1.0"}
    }
}

# call get_current_time tool
tool_call_request = {
    "jsonrpc": "2.0",
    "id": 2,
    "method": "tools/call",
    "params": {
        "name": "get_current_time",
        "arguments": {
            "timezone": "Europe/Rome"
        }
    }
}

process.stdin.write(json.dumps(init_request) + '\n')
process.stdin.write(json.dumps(tool_call_request) + '\n')
process.stdin.flush()
process.stdin.close()

# Read responses
for line in process.stdout:
    if line.strip():
        response = json.loads(line)
        print(json.dumps(response, indent=2))
````
