```python
# if npx is not found, check nvm installation
# then add nvm path for this notebook
import os
import platform
import glob

# Find the latest node version or use a wildcard 
path = f'%APPDATA%\\nvm' if platform.system() == 'Windows' else f'/home/{os.getenv('USER') or os.getenv('USERNAME')}/.nvm/versions/node/*/bin'
if os.path.exists(path):
    nvm_node_bins = glob.glob(path)
    if nvm_node_bins:
        # Use the first match (or latest if sorted)
        node_bin = sorted(nvm_node_bins)[-1]
        os.environ['PATH'] = f"{node_bin}:{os.environ['PATH']}"
        print(f"✓ Added to PATH: {node_bin}")
    else:
        print("⚠ NVM node installation not found")
else:
    print("⚠ NVM path does not exist")
    print("check on windows with psw:  $env:NVM_HOME")
    print("check on linux/macOs: echo $NVM_INC or echo $NVM_DIR")
    print("To install nvm: https://github.com/nvm-sh/nvm") 
```
