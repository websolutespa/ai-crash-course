```python
from langchain_ollama import ChatOllama
#creative task
model = "gpt-oss:20b" #choose the right model => general purpose model
def _generate(messages):
    llm = ChatOllama(
        model=model,
        reasoning=False, #need reasoning for this task? 
        temperature=0.7, #set randomness to improve creativity, if needed
        num_predict=2048, #give necessary space for task response
        )
    rs = llm.invoke(messages)
    print(f"{rs.content}")
    return rs.content
```

```python
#naive prompt
prompt = "Brainstorm a list of product names for shoes that fits any foot size."
messages = [("user", prompt)]
response = _generate(messages)
```

```python
# prompt template with variables

from langchain_ollama import ChatOllama
#creative task
model = "gpt-oss:20b" #choose the right model => general purpose model
def _generate(messages):
    llm = ChatOllama(
        model=model,
        reasoning=False, #need reasoning for this task? 
        temperature=1.0, #set randomness to improve creativity, if needed
        num_predict=1024, #give necessary space for task response
        )
    rs = llm.invoke(messages)
    print(f"{rs.content}")
    return rs.content

PROMPT_TEMPLATE = """"You are a creative consultant brainstorming names for businesses.
Create catchy and memorable names for start-up companies, associated with a unique claim for each name.

You must follow the following principles:
- Each name should be short and easy to remember. 
- Each name should be easy to pronounce. 
- Each name should be unique and not already taken by another company.
- Each claim must be aligned with the company solution and industry.
- No bold or italic formatting, no backticks, no markdown, just plain text.

Example of output format:
1. name1: claim1
2. name2: claim2
3. name3: claim3
4. name4: claim4
5. name5: claim5

Please generate a numerical list of {number} catchy names for a start-up in the `{industry}` industry that deals with `{solution}`
"""


companies = [
    {"industry": "medical", "solution": "creating AI solutions by automatically summarizing patient records"},
    {"industry": "fashion", "solution": "sustainable clothing made from recycled materials"},
    {"industry": "food and beverage", "solution": "healthy snacks for busy professionals"},
]
for company in companies:
    print(f"\n{company['industry']} / {company['solution']}\n")
    prompt = PROMPT_TEMPLATE.format(industry=company["industry"], solution=company["solution"], number=5)
    messages= [("user", prompt)]
    _generate(messages)
```
