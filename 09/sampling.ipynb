```python
from langchain_ollama import ChatOllama

#sampling parameters
samplings = [
    {"temperature": 0, "top_k": 0}, #deterministic
    {"temperature": 0.5, "top_k": 500}, #balanced
    {"temperature": 5, "top_k": 2000}, #creative    
    ]

def _generate(model, prompt):
    for s in samplings:
        llm = ChatOllama(
        model=model,
        reasoning=False,
        temperature=s["temperature"],
        top_k=s["top_k"]
        )
        print(f"Sampling params: temp={s['temperature']}, top_k={s['top_k']}")
        for i in range(3):
            rs = llm.invoke([
                ("user", prompt),
            ])        
            print(f"{i+1}. {rs.content}")
        print("=================================\n")
```

```python
# creative generative task
model = "granite4:3b"
prompt = "Write an haiku on prompt sampling parameters."
_generate(model, prompt)
```

```python
# factual task
model = "granite4:3b"
prompt = "Who won the FIFA World Cup in 2018?"
_generate(model, prompt)
```

```python
# math task
model = "granite4:3b"
prompt = "What is the sum of the first 50 positive integers? Return only the number." #note how no explanations reduced randomness
_generate(model, prompt)
```

```python
#max tokens truncation test
model = "granite4:3b"
prompt = "List the first 20 US presidents."
llm = ChatOllama(
    model=model,
    reasoning=False,
    num_predict=100 #too small to fit the full answer
    )
rs = llm.invoke([
    ("user", prompt),
])
print(f"{rs.content}")
```

```python
#stop sequences behavior test
model = "granite4:3b"
prompt = "List the first 20 US presidents."

llm = ChatOllama(
    model=model,
    reasoning=False,
    stop=["\n"] #stop at the first newline
    )
rs = llm.invoke([
    ("user", prompt),
])
print(f"{rs.content}")

print("----")

llm = ChatOllama(
    model=model,
    reasoning=False,
    stop=["15."] #stop before the 15th president
    )
rs = llm.invoke([
    ("user", prompt),
])
print(f"{rs.content}")
```

```python
#not all models respond the same way to the same prompt
prompt = "Give a brief description of a python"
def _generate(_model, prompt):
    llm = ChatOllama(
        model=_model,
        reasoning=False,
        temperature=0,
        )
    rs = llm.invoke([
        ("user", prompt),       
    ])
    print(f"{rs.content}")

_generate("gpt-oss:20b", prompt) #general purpose model
print("----")
_generate("qwen3-coder:30b", prompt) #more capable model but fine-tuned for specific domain
```
