```python
from langchain_openai import ChatOpenAI

def _generate(prompt,presence_penalty=1.0):
    llm = ChatOpenAI(
    model="gpt-4.1",
    temperature=0.5,
    presence_penalty=presence_penalty
    )
    rs = llm.invoke([("user", prompt),]) 
    print(rs.content[:100])
    print("\nTop 10 words in the response:")
    _top_words(rs.content)

def _top_words(text):
    from collections import Counter
    import re
    words = re.findall(r'\w+', text.lower())
    counter = Counter(words)
    most_common = counter.most_common(10)
    for word, count in most_common:
        print(f"{word}: {count}")
```

```python
# show how different repeat penality affects the output
prompt = "What features make a dog a good dog?"
for rp in [0.5, 1.2, 2.0]:
    print(f"Presence Penalty: {rp}")
    _generate(prompt, presence_penalty=rp)
    print("=================================\n")
```
