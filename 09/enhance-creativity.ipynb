```python
from langchain_core.messages import HumanMessage, SystemMessage
from langchain.chat_models import init_chat_model

#model = "gpt-oss:20b"
model = "ollama:granite4:3b"
llm = init_chat_model(
    model=model,
    temperature=1, # improve creativity with higher temperature
    timeout=30,
    max_tokens=1000,
)
```

```python
# basic: lack of diversity (same response repeated or similar in pattern), preamble not requested
system_prompt = """You are a helpful assistant."""
user_query = "Tell me a joke about coffee."
for _ in range(10):
    response = await llm.ainvoke(
        [
            SystemMessage(content=system_prompt),
            HumanMessage(content=user_query),
        ]
    )
    print(response.content)
    print("-----")
```

```python
#verbalized sampling 
vs_system_prompt = """
You are a helpful assistant.

For each query, please generate a set of five possible responses, each within a separate <response> tag. 
Responses should each include a <text> and a numeric <probability>. 
Sample at random from the [full distribution / tails of the distribution, such that the probability of each response is less than 0.10]
"""
user_query = "Tell me a joke about coffee."
response = await llm.ainvoke(
    [
        SystemMessage(content=vs_system_prompt),
        HumanMessage(content=user_query),
    ]
)
print(response.content)

# even better with proper role
vs_system_prompt ="""
You are a creative assistant that helps users generate creative ideas and solutions. 
Use your imagination and think outside the box to provide unique and original responses.

For each query, please generate a set of five possible responses, each within a separate <response> tag. 
Responses should each include a <text> and a numeric <probability>. 
Sample at random from the [full distribution / tails of the distribution, such that the probability of each response is less than 0.10]
"""
```

Reverse prompt

```python
i_like_this = """
At its core, the art of prompting is about bridging the gap between intention and action, purpose and consequence. It requires a deep understanding of our audience's needs, desires, and aspirations, as well as an appreciation for the nuances of language, culture, and context. By honing this skill set, we can create prompts that not only capture attention but also inspire trust, empathy, and connection. Whether you're crafting copy for a marketing campaign or designing an interactive experience, remember that every prompt is an invitation to explore new possibilities together.
"""
```

```python
#one-shot
user_query = f"""
Write an intro for a Medium article about why prompts matter.

Here is an example of a good intro:
{i_like_this}
"""
response = await llm.ainvoke(
    [
        SystemMessage(content=system_prompt),
        HumanMessage(content=user_query),
    ]
)
print(response.content)
```

```python
#reverse prompt
rp_system_prompt = """
You are an expert prompt engineer, specializing in reverse prompting engineering.
Given the user query, generate a prompt that would lead to that response.
Output only the prompt, without any additional text.

Reverse prompt this:
"""

user_query = """
{i_like_this}
"""
response = await llm.ainvoke(
    [
        SystemMessage(content=rp_system_prompt.format()),
        HumanMessage(content=user_query.format(i_like_this=i_like_this)),           
    ]
)
print(response.content)

print("---- running the generated prompt ----")

new_prompt = response.content
response = await llm.ainvoke(
    [
        SystemMessage(content=system_prompt),
        HumanMessage(content=new_prompt),           
    ]
)
print(response.content)
```

```python
#reverse prompt + scope/context
rp_system_prompt = """
You are an expert prompt engineer, specializing in reverse prompting engineering.
Given the user query, generate a prompt that would lead to that response.
Output only the prompt, without any additional text.

User scope: {scope}

Reverse prompt this:
"""

user_query = """
{i_like_this}
"""
response = await llm.ainvoke(
    [
        SystemMessage(content=rp_system_prompt.format(scope="Write an intro for a Medium article about why prompts matter.")),
        HumanMessage(content=user_query.format(i_like_this=i_like_this)),           
    ]
)
print(response.content)

print("---- running the generated prompt ----")

new_prompt = response.content
response = await llm.ainvoke(
    [
        SystemMessage(content=system_prompt),
        HumanMessage(content=new_prompt),           
    ]
)
print(response.content)

# can be applied to writing, summary, explanation etc...
```
