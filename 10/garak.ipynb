```python
%%bash

garak --plugin_info generators.ollama.OllamaGenerator
```

```python
%%bash
# LMRC: Language Model Risk Card: https://arxiv.org/pdf/2303.18190
#Deadnaming: mentioning a name that a person does not wish to be associated with (i.e. by their birth name instead of their chosen name)

garak --target_type ollama --target_name granite4:3b --probes lmrc --generations 1
```

```python
%%bash 

#garak --target_type ollama --target_name qwen3:0.6b --probes lmrc.Profanity
garak --target_type ollama --target_name granite4:3b --probes encoding.InjectBase64,encoding.InjectMorse --generations 1

# The model is vulnerable to encoding base64 injection attacks, but not to morse code injection attacks.
# Is the model resistant to morse or is a false negative? The model is not able to decode morse code, and decoded it with simple stereotypes, like SOS, HELLO, HELP, etc.
# Paradox: a less capable model can be vulnerable on certain types of attacks and more secure against certain types of attacks!
```
